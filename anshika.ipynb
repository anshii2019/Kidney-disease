#Strategy for Handling Missing Values

1. Drop rows with missing values (if the dataset is large).
2. Fill missing values using the mean, median, or mode.
3. Fill missing values with a placeholder (like "Unknown" for categorical data)

```python
# Filling missing values for numerical columns using median
# 1.numerical data(continuous)
#median- (if outliers are present
#mode- (if no significant outliers)
df['age'] = df['age'].fillna(df['age'].median())
df['bp'] = df['bp'].fillna(df['bp'].median())
df['sg'] = df['sg'].fillna(df['sg'].median())
df['al'] = df['al'].fillna(df['al'].median())
df['su'] = df['su'].fillna(df['su'].median())
df['bgr'] = df['bgr'].fillna(df['bgr'].median())
df['bu'] = df['bu'].fillna(df['bu'].median())
df['sc'] = df['sc'].fillna(df['sc'].median())
df['hemo'] = df['hemo'].fillna(df['hemo'].median())
#df['rc'] = df['rc'].fillna(df['rc'].median()) wc,pcv,

# Filling missing values for numerical columns using mean
df['sod'] = df['sod'].fillna(df['sod'].mean())
df['pot'] = df['pot'].fillna(df['pot'].mean())

# 2. Categorical Data (Discrete)
# Filling missing values for categorical columns using mode
df['rbc'] = df['rbc'].fillna(df['rbc'].mode()[0])
df['pc'] = df['pc'].fillna(df['pc'].mode()[0])
df['pcc'] = df['pcc'].fillna(df['pcc'].mode()[0])
df['ba'] = df['ba'].fillna(df['ba'].mode()[0])
df['htn'] = df['htn'].fillna(df['htn'].mode()[0])
df['dm'] = df['dm'].fillna(df['dm'].mode()[0])
df['cad'] = df['cad'].fillna(df['cad'].mode()[0])
df['appet'] = df['appet'].fillna(df['appet'].mode()[0])
df['pe'] = df['pe'].fillna(df['pe'].mode()[0])
df['ane'] = df['ane'].fillna(df['ane'].mode()[0])
```

Converting Data Types -
Some numerical values are stored as objects (text), so we need to convert them.

```python
#Convert Numerical Columns to Float
df['pcv'] = pd.to_numeric(df['pcv'], errors='coerce')  
df['wc'] = pd.to_numeric(df['wc'], errors='coerce')  
df['rc'] = pd.to_numeric(df['rc'], errors='coerce')  

#pd.to_numeric(df['pcv'], errors='coerce') → Converts object (text) values to numeric (float).
#errors='coerce' → If a value cannot be converted, it becomes NaN
```

```python
print(df.isnull().sum())
```

Convert Binary Categorical Columns to 0/1 --Many categorical values are Yes/No and can be converted into 0 and 1 for easier analysis.

```python
binary_columns = ['htn', 'dm', 'cad', 'pe', 'ane']
df[binary_columns] = df[binary_columns].apply(lambda x: x.map({'yes': 1, 'no': 0}))
```

✅ Convert Target Variable (classification) to 0/1

```python
df['classification'] = df['classification'].map({'ckd': 1, 'notckd': 0})
```

Removing Duplicate Entries

```python
df.drop_duplicates(inplace=True)
```

1. Removes spaces
2. Converts to lowercase

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
# For numerical columns, use mean or median:
df['rc'] = df['rc'].fillna(df['rc'].median())
```

Normalizing Numerical Data
-- Features like bp (blood pressure), age, wc, rc have different ranges. Scaling makes all values comparable.

```python
# Min-Max scaling brings values between 0 and 1
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['age', 'bp', 'rc']] = scaler.fit_transform(df[['age', 'bp', 'rc']])
```

```python
# Standardization (Mean = 0, Std Dev = 1)
# For algorithms like logistic regression, we use standardization:

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['age', 'bp', 'rc']] = scaler.fit_transform(df[['age', 'bp', 'rc']])
```
